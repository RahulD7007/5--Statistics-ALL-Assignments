#### 1. Basic Questions:-
1. **What is histogram?** -> A histogram is a bar chart for continuous data where bars represent frequency (or density) of values falling into contiguous intervals (bins). Unlike a bar chart, bars touch each other — because continuous data has no gaps.
2. **What is a bin?** -> A bin is a fixed-width interval that groups continuous data values.
3. **What does frequency mean?** ->
    - **Count frequency** = how many data points fall in that bin.
    - **Relative frequency** = count / total (gives probability interpretation).
    - Density = relative frequency / bin width (area under curve = 1)
4. **Why are histograms useful?** -> They reveal the **shape of a distribution** — which determines which statistics and models are valid:
> 
> | Shape | What it means | ML implication |
> |-------|--------------|----------------|
> | Bell-curve | Normal | Many algorithms assume this |
> | Right-skewed | Outliers pull mean right | Log-transform before modeling |
> | Bimodal | Two subgroups | Consider clustering first |
> | Uniform | No dominant range | Check if data is valid |

#### 2. Numerical Problems:-
---
#### Given the data: 12, 15, 18, 20, 22, 25, 27, 30
---
1. Create bins of size 10.
2. Find the frequency in each bin

--- 
#### Given the scores: 45, 50, 55, 60, 65, 70, 75, 80
---
1. Create a histogram.
2. Identify the most common score range.

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# ── 1. Create bins of size 10 ─────────────────────────────
data1 = [12, 15, 18, 20, 22, 25, 27, 30]
bins1 = [10, 20, 30, 40]  # bin size = 10

# ── 2. Find the frequency in each bin ─────────────────────────────
labels = ['[10–20)', '[20–30)', '[30–40)']
freqs  = [
    sum(10 <= x < 20 for x in data1),
    sum(20 <= x < 30 for x in data1),
    sum(30 <= x <= 40 for x in data1)
]
print("Bin Frequencies:")
for l, f in zip(labels, freqs):
    print(f"  {l}: {f} values  {'█'*f}")
```

```python
# ── 1. Create & Scores a histogram. ─────────────────────────────

scores = [45, 50, 55, 60, 65, 70, 75, 80]

fig, axes = plt.subplots(1, 2, figsize=(13, 4))

axes[0].hist(data1, bins=bins1, color='steelblue', edgecolor='white', rwidth=0.95)
axes[0].set_title('Data Distribution (bin size=10)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Value'); axes[0].set_ylabel('Frequency')
for i, (l, f) in enumerate(zip(labels, freqs)):
    axes[0].text(10 + i*10 + 5, f + 0.05, str(f), ha='center', fontweight='bold')

n, bins_out, patches = axes[1].hist(scores, bins=5, color='coral', edgecolor='white', rwidth=0.95)
axes[1].set_title('Exam Score Distribution', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Score'); axes[1].set_ylabel('Frequency')

# ── 2. Identify the most common score range. ─────────────────────────────
max_idx = np.argmax(n)
patches[max_idx].set_facecolor('gold')
patches[max_idx].set_edgecolor('darkorange')
axes[1].legend(handles=[mpatches.Patch(color='gold', label=f'Most common: [{bins_out[max_idx]:.0f}–{bins_out[max_idx+1]:.0f}]')])

plt.tight_layout()
plt.show()
print(f"\nMost common score range: [{bins_out[max_idx]:.0f} – {bins_out[max_idx+1]:.0f})")
```

#### 3. Conceptual Questions:-
1. How is a histogram different from a bar chart?
2. What happens if the bin size is too large?
3. What happens if the bin size is too small?
4. How can a histogram help identify outliers?

**Q1. How is a histogram different from a bar chart?**

| Feature | Histogram | Bar Chart |
|---------|-----------|----------|
| Data type | Continuous (numerical) | Categorical (discrete) |
| Bars touch? | Yes (no gaps in data) | No (categories are separate) |
| X-axis | Numerical ranges (bins) | Category labels |
| Purpose | Show distribution shape | Compare category values |
| Reordering bars | Breaks meaning | Fine (no natural order) |

> **Note:** "In a histogram, bar order is mathematically enforced by the number line. In a bar chart, bar order is arbitrary — you could sort by height for readability."

---

**Q2. What happens if bin size is too large?**
> **Over-smoothing** — you lose detail. A bimodal distribution can look unimodal. You can miss the structure entirely. Think of it as zooming too far out on a map.

> **Note:-** Large bins = high bias, low variance. 

**Q3. What happens if bin size is too small?**
> **Under-smoothing (noisy)** — every data point gets its own bin. The histogram looks like random spikes. No pattern is visible. Equivalent to memorizing noise.

> **Note:-** Small bins = low bias, high variance. 

**Q4. How can a histogram help identify outliers?**
> Outliers appear as **isolated bars far from the main mass** — a bar separated by empty bins. This is actually more visually intuitive than a Z-score for understanding *where* in the distribution the outlier lives.

#### 1. Basic Questions:-
1. What is a Box Plot?
2. What are the five values used in a Box Plot?

**Q1. What is a Box Plot?**
> A box plot compresses an entire distribution into **5 numbers**: Min, Q1, Median, Q3, Max — plus a rule-based definition of outliers. It sacrifices granularity (you can't see shape) for the ability to **compare multiple distributions side-by-side** instantly.

**Q2. Five values in a Box Plot:**

| # | Name | Meaning |
|---|------|---------|
| 1 | Minimum | Smallest non-outlier value |
| 2 | Q1 (25th percentile) | 25% of data is below this |
| 3 | Median (Q2) | Middle value — robust to outliers |
| 4 | Q3 (75th percentile) | 75% of data is below this |
| 5 | Maximum | Largest non-outlier value |

**Q3. What does the median represent?**
> The value that splits data exactly 50/50. Unlike the mean, it is **resistant to outliers** — one extreme value cannot move it more than one rank. That's why income distribution always uses median, not mean.

**Q4. Purpose of whiskers?**
> Whiskers extend to the last data point within **1.5 × IQR** from Q1/Q3. Anything beyond whiskers = outlier (plotted as individual dots). The 1.5 multiplier was chosen by Tukey empirically — it flags ~0.7% of normal data as outliers, which is a sensible threshold.

#### 2. Numerical Problems:-
1. Find the Five-Number Summary for: 10, 12, 15, 18, 20, 25, 28, 30, 35
2. Calculate the IQR for: 5, 7, 9, 10, 12, 15, 18, 20, 25
3. Determine whether 100 is an outlier in: 10, 12, 14, 16, 18, 20, 22, 24, 100

```python
import numpy as np
import matplotlib.pyplot as plt

# ── Q1: Find the Five-Number Summary ────────────────────────
data_a = sorted([10, 12, 15, 18, 20, 25, 28, 30, 35])
n = len(data_a)
Q1 = np.percentile(data_a, 25)
Q2 = np.percentile(data_a, 50)  # median
Q3 = np.percentile(data_a, 75)
print("── Five-Number Summary ──")
print(f"  Min    = {min(data_a)}")
print(f"  Q1     = {Q1}")
print(f"  Median = {Q2}")
print(f"  Q3     = {Q3}")
print(f"  Max    = {max(data_a)}")
```

```python
# ── Q2: Calculate the IQR ────────────────────────────────────────
data_b = [5, 7, 9, 10, 12, 15, 18, 20, 25]
q1_b = np.percentile(data_b, 25)
q3_b = np.percentile(data_b, 75)
IQR = q3_b - q1_b
print(f"\n── IQR Calculation ──")
print(f"  Q1={q1_b}, Q3={q3_b}")
print(f"  IQR = Q3 - Q1 = {q3_b} - {q1_b} = {IQR}")
```

```python
# ── Q3: Is 100 an outlier? ─────────────────────────
data_c = [10, 12, 14, 16, 18, 20, 22, 24, 100]
q1_c = np.percentile(data_c, 25)
q3_c = np.percentile(data_c, 75)
iqr_c = q3_c - q1_c
upper_fence = q3_c + 1.5 * iqr_c
lower_fence = q1_c - 1.5 * iqr_c
print(f"\n── Outlier Detection (value=100) ──")
print(f"  Q1={q1_c}, Q3={q3_c}, IQR={iqr_c}")
print(f"  Upper fence = Q3 + 1.5×IQR = {q3_c} + {1.5*iqr_c} = {upper_fence}")
print(f"  100 > {upper_fence}? → {'YES — 100 is an outlier!' if 100 > upper_fence else 'No'}")
```

```python
# ── Visualization ─────────────────────────────────────────
fig, axes = plt.subplots(1, 3, figsize=(14, 5))

for ax, d, title in zip(axes,
    [data_a, data_b, data_c],
    ['Five-Number Summary', 'IQR Calculation', 'Outlier Detection (100)']):
    bp = ax.boxplot(d, vert=True, patch_artist=True,
                    boxprops=dict(facecolor='steelblue', alpha=0.6),
                    medianprops=dict(color='red', linewidth=2),
                    flierprops=dict(marker='o', color='crimson', markersize=10))
    ax.set_title(title, fontweight='bold')
    ax.set_ylabel('Value')

plt.tight_layout()
plt.show()
```

#### 3. Conceptual Questions:- 
1. Why is IQR useful?
2. Why are Box Plots good for detecting outliers?
3. What information does the box represent?
4. What does a long whisker indicate?

**Q1. Why is IQR useful?**
> IQR = middle 50% of data. It is completely **immune to outliers** — no matter how extreme the top 25% or bottom 25% values are, IQR doesn't change. Standard deviation, by contrast, gets inflated by every outlier.
> 
> **Note**: Use IQR when data is skewed or has outliers. Use standard deviation only when data is approximately normal.

**Q2. Why are Box Plots good for detecting outliers?**
> The 1.5 × IQR rule is automatic, objective, and distribution-free. No assumption of normality needed (unlike Z-score which assumes normal distribution). Any point beyond the whiskers is flagged mathematically.

**Q3. What information does the box represent?**
> The box = **IQR = middle 50% of data**. The width of the box = spread of the 'typical' values. A narrow box = data is concentrated. A wide box = high variability among typical values.

**Q4. What does a long whisker indicate?**
> High spread in that tail — data is spread out in that direction. A long right whisker with a short left whisker = right skew. **Asymmetric whiskers = asymmetric distribution.**

#### 1. Basic Questions:-

#### **Q1. What is a scatter plot?**
> A 2D plot where each point = one observation with two attributes. It visually answers: *Is there a relationship between variable X and variable Y?*
> 
> Scatter plot is the **first tool you run before any regression** — it tells you if linear regression is even appropriate.

#### **Q2. What does each point represent?**
> One individual observation (row in your dataset) with two features plotted on X and Y axes. In a dataset of 1000 students, each dot = one student's (hours_studied, exam_score) pair.

#### **Q3. What is correlation?**
> The strength and direction of the **linear** relationship between two variables. Pearson r ranges from -1 to +1.

| r value | Interpretation |
|---------|----------------|
| +0.8 to +1.0 | Strong positive |
| +0.4 to +0.8 | Moderate positive |
| -0.4 to +0.4 | Weak/no linear relationship |
| -0.8 to -0.4 | Moderate negative |
| -1.0 to -0.8 | Strong negative |

> **Note:** r = 0 does NOT mean no relationship. It means no *linear* relationship. A perfect U-shape gives r ≈ 0. Always look at the scatter plot, never just the number.

#### **Q4. What is an outlier?**
> A point that doesn't follow the overall pattern — sits far from the main cluster. In scatter plots, it can: (1) distort the correlation coefficient, (2) represent data entry error, (3) represent a genuinely interesting case worth investigating.

#### 2. Numerical Problems:-
1. Plot the following data using a Scatter Plot:
    - Hours Studied: 1, 3, 5, 7, 9
    - Scores: 30, 50, 65, 78, 95
2. Identify whether the relationship is positive or negative.
3. Find any outliers in the dataset.

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

hours  = [1, 3, 5, 7, 9]
scores = [30, 50, 65, 78, 95]

# Correlation
r, p_val = stats.pearsonr(hours, scores)
slope, intercept, _, _, _ = stats.linregress(hours, scores)
x_line = np.linspace(0, 10, 100)
y_line = slope * x_line + intercept

fig, ax = plt.subplots(figsize=(8, 5))
ax.scatter(hours, scores, color='steelblue', s=120, zorder=5, label='Observations')
ax.plot(x_line, y_line, 'r--', linewidth=2, label=f'Trend line (r={r:.3f})')

for h, s in zip(hours, scores):
    ax.annotate(f'({h},{s})', (h, s), textcoords='offset points', xytext=(8, 4), fontsize=9)

ax.set_xlabel('Hours Studied', fontsize=12)
ax.set_ylabel('Exam Score', fontsize=12)
ax.set_title('Hours Studied vs Exam Score', fontsize=13, fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"Pearson r = {r:.4f}  → Strong positive correlation")
print(f"Regression: Score = {slope:.2f}×Hours + {intercept:.2f}")
print(f"p-value = {p_val:.4f}  → {'Statistically significant' if p_val < 0.05 else 'Not significant'}")
print("Outliers: None — all points follow the trend closely")
```

#### 3. Conceptual Questions:-

2. What is the difference between positive and negative correlation?

#### **Q1. Why are Scatter Plots useful?**
> They reveal **4 things no summary statistic can**: direction of relationship, strength, linearity (or non-linearity), and presence of outliers. Anscombe's Quartet famously shows 4 datasets with identical mean, std, and correlation — but completely different scatter plots.

#### **Q2. What is the difference between positive and negative correlation?**

| | Positive | Negative |
|--|---------|----------|
| Pattern | Points go up-right | Points go down-right |
| Meaning | X↑ → Y↑ | X↑ → Y↓ |
| Example | Study hours vs Score | Stress vs Sleep quality |

#### **Q3. Why should outliers be investigated?**
> Never delete outliers without investigation. They can be: (1) **Data errors** → fix/remove, (2) **Legitimate extremes** → keep and note, (3) **The most important signal** → fraud detection, rare disease cases. In ML, one outlier can shift a regression line dramatically.

#### **Q4. Can a Scatter Plot prove causation?**
> **No.** Correlation ≠ Causation. Classic example: ice cream sales and drowning both rise in summer. The actual cause is heat, not ice cream. To prove causation, you need a **controlled experiment (A/B test)** or **causal inference methods** (instrumental variables, DAGs).

#### 1. Basic Questions:-
1. What is probability?
2. What is a sample space?
3. What is an event?
4. What is conditional probability?I
5. What is Bayes' Theorem?
6. What is a random variable?

#### **Q1. What is probability?**
> A number between 0 and 1 that quantifies our **degree of belief** that an event will occur. P=0 means impossible; P=1 means certain; P=0.5 means equally likely either way.

#### **Q2. What is a sample space?**
> The set of **all possible outcomes** of an experiment. Everything that *could* happen. Probability of the entire sample space = 1 (something must happen).

#### **Q3. What is an event?**
> A **subset of the sample space** — one or more outcomes we care about. Rolling a die: S = {1,2,3,4,5,6}. Event 'even number' = {2,4,6}. P(event) = |event| / |sample space| = 3/6 = 0.5.

#### **Q4. What is conditional probability?**
> P(A|B) = Probability of A *given that* B has already occurred. B reduces the effective sample space.
> 
> Formula: **P(A|B) = P(A∩B) / P(B)**
> 
> Example: P(person has disease | test is positive) — this is the clinically useful probability, not P(positive test | has disease).


#### **Q5. What is Bayes' Theorem?**

> A formula to **update beliefs** when new evidence arrives:
> 
> **P(H|E) = P(E|H) · P(H) / P(E)**
> 
> | Term | Name | Meaning |
> |------|------|---------|
> | P(H) | Prior | Belief before seeing evidence |
> | P(E\|H) | Likelihood | How probable is evidence if H is true |
> | P(H\|E) | Posterior | Updated belief after evidence |
> | P(E) | Marginal | Total probability of evidence |
> 
> **Bayes = learning from evidence mathematically**. The prior is what you knew before. The posterior is what you know now.

#### **Q6. What is a random variable?**
> A **function** that maps outcomes of a random experiment to numbers. Not actually 'random' or a 'variable' in the traditional sense — it's a mathematical bridge between outcomes and numbers we can compute with.

#### 3. Conceptual Questions:-
1. Why is sample space important?
2. How is conditional probability different from regular probability?
3. Why do doctors use Bayes' Theorem?
4. What is the difference between discrete and continuous random variables?

#### **Q1. Why is sample space important?**
> You cannot calculate probability without knowing all possible outcomes. An incorrect sample space leads to incorrect probabilities. It defines the **denominator** of every probability calculation.

#### **Q2. How is conditional probability different from regular probability?**
> Regular probability: P(A) — no prior knowledge assumed.  
> Conditional: P(A|B) — B has occurred, which changes what's possible.  
> **Key insight:** P(A|B) can be vastly different from P(A). P(fire | smoke) >> P(fire) because smoke is strong evidence.

#### **Q3. Why do doctors use Bayes' Theorem?**
> A test might have 99% sensitivity (P(positive|disease)=0.99). But if the disease is rare (P(disease)=0.001), then a positive test might only mean P(disease|positive) ≈ 9%. Without Bayes, doctors would over-diagnose. This is the **base rate fallacy** — Bayes' theorem prevents it.

#### **Q4. What is the difference between discrete and continuous random variables?**

| Feature | Discrete | Continuous |
|---------|----------|------------|
| Values | Countable (0,1,2...) | Uncountable (any real number) |
| Probability tool | PMF (mass) | PDF (density) |
| Example | Die roll, coin flips | Height, temperature, time |
| P(X = exact value) | Can be > 0 | Always = 0 |
| Cumulative | Summation Σ | Integration ∫ |

#### **Q1. How is probability used in weather forecasting?**
> Meteorologists run thousands of simulations with slightly different initial conditions. The percentage of simulations where rain occurs = P(rain). It's frequentist probability computed over ensemble models, not a guess.

#### **Q2. How does Netflix use probability in recommendations?**

#### 1. Basic Questions:-
1. What is a probability distribution?
2. What is a random variable?
3. What is the difference between PMF and PDF?
4. What is a CDF?

#### **Q1. What is a probability distribution?**
> A complete description of all possible values a random variable can take **and** how probable each value is. It answers: *what's the full landscape of outcomes?*

#### **Q2. What is a random variable?**
> A **function** that maps outcomes of a random experiment to numbers. Not actually 'random' or a 'variable' in the traditional sense — it's a mathematical bridge between outcomes and numbers we can compute with.

#### **Q3. What is the difference between PMF and PDF?**

| Feature | PMF (Probability Mass Function) | PDF (Probability Density Function) |
|---------|--------------------------------|-------------------------------------|
| For | Discrete r.v. | Continuous r.v. |
| Output | P(X = x) directly (a probability) | Density — NOT probability |
| Property | All values sum to 1: Σ P(x)=1 | Area under curve = 1: ∫f(x)dx=1 |
| P(X = exact value) | Can be > 0 | Always = 0 |
| Example | Die roll, binomial | Normal, exponential |

> **Critical insight:** For continuous distributions, P(X = exactly 1.5) = 0. You can only ask P(1.4 < X < 1.6) = area under PDF. Interviewers love asking why.

#### **Q4. What is a CDF?**
> CDF F(x) = P(X ≤ x) — the probability that the variable takes a value **at or below x**.
> 
> CDF is always monotonically non-decreasing from 0 to 1. **Derivative of CDF = PDF.** CDF is far more practically useful than PDF for answering 'what fraction of people scored below 80?'

#### 2. Numerical Problems:-
1. A fair coin is tossed once. Find the probability of getting Head.
2. A fair dice is rolled once. Find P(X=4).
3. A coin is tossed 5 times. Find the probability of getting exactly 3 Heads.

```python
from scipy.stats import binom
from scipy import stats
import numpy as np
import matplotlib.pyplot as plt
from math import comb

# Q1: P(Head) on fair coin
print("── Q1: P(Head) fair coin ──")
print("  P(Head) = 1/2 = 0.5")

# Q2: P(X=4) on die
print("\n── Q2: P(X=4) on die ──")
print("  P(X=4) = 1/6 ≈ 0.1667")

# Q3: Binomial — P(exactly 3 Heads in 5 tosses)
n, k, p = 5, 3, 0.5
prob = binom.pmf(k, n, p)
manual = comb(n, k) * (p**k) * ((1-p)**(n-k))

print(f"\n── Q3: P(exactly 3 Heads in 5 tosses) ──")
print(f"  Formula: C(5,3) × (0.5)³ × (0.5)²")
print(f"  = {comb(5,3)} × {0.5**3} × {0.5**2}")
print(f"  = {manual:.4f}")
print(f"  scipy verification: {prob:.4f}")

# Visualization: Binomial PMF
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

k_vals = np.arange(0, 6)
pmf_vals = binom.pmf(k_vals, 5, 0.5)
colors_pmf = ['gold' if k == 3 else 'steelblue' for k in k_vals]
axes[0].bar(k_vals, pmf_vals, color=colors_pmf, edgecolor='white')
axes[0].set_title('PMF: Binomial(n=5, p=0.5)', fontweight='bold')
axes[0].set_xlabel('Number of Heads (k)')
axes[0].set_ylabel('P(X = k)')
for k, v in zip(k_vals, pmf_vals):
    axes[0].text(k, v + 0.005, f'{v:.3f}', ha='center', fontsize=9)
axes[0].legend(handles=[plt.Rectangle((0,0),1,1,color='gold')], labels=['k=3 (our answer)'])

# CDF
cdf_vals = binom.cdf(k_vals, 5, 0.5)
axes[1].step(k_vals, cdf_vals, where='post', color='coral', linewidth=2.5)
axes[1].scatter(k_vals, cdf_vals, color='coral', s=80, zorder=5)
axes[1].set_title('CDF: Binomial(n=5, p=0.5)', fontweight='bold')
axes[1].set_xlabel('k'); axes[1].set_ylabel('P(X ≤ k)')
axes[1].grid(True, alpha=0.3)
for k, v in zip(k_vals, cdf_vals):
    axes[1].text(k+0.05, v-0.05, f'{v:.3f}', fontsize=9)

plt.tight_layout()
plt.show()
```

#### 3. Conceptual Questions:-
1. Why is PDF used for continuous data?
2. Why is the normal distribution important?
3. What conditions are required for a binomial distribution?

#### **Q1. Why is PDF used for continuous data?**
> For continuous r.v., the probability of any exact value is 0 (infinitely many possible values). PDF gives **density** — probability per unit of measurement. To get actual probability, integrate between two bounds. This is why PDF(x) can exceed 1 — it's density, not probability.

#### **Q2. Why is the normal distribution important?**
> Three reasons:
> 1. **Central Limit Theorem** — sample means of *any* distribution converge to normal as n→∞
> 2. **Nature** — heights, measurement errors, test scores naturally approximate it
> 3. **Math convenience** — sum of normals is normal; linear combinations stay tractable
> 
> The normal distribution is not "common in nature" — it's common in *sample averages*. This distinction is crucial.

#### **Q3. What conditions are required for a binomial distribution?**
| Condition | What it means |
|-----------|---------------|
| Fixed n | Set number of trials |
| Binary outcomes | Only 2 outcomes per trial (Success/Fail) |
| Constant p | Same probability of success each trial |
| Independence | Each trial doesn't affect others |

> Break any condition → binomial is wrong. (e.g., drawing cards without replacement violates independence → use Hypergeometric instead)

#### 4. Real-world Application Questions:-
1. How can probability distributions help in weather forecasting?
> Daily rainfall → **Exponential distribution** (time between rain events). Temperature → **Normal distribution** (mean + variance shift by season). Extreme events → **Generalized Extreme Value (GEV) distribution**. The choice of distribution determines the model and its predictions.

>  Each distribution is a different *shape assumption* about the world. Choosing the wrong distribution = systematic forecasting error.
